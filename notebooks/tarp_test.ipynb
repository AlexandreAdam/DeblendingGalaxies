```python
import numpy as np
import matplotlib.pyplot as plt
import tarp
plt.style.use("science")
```

```python
D = 1000
N = 100

t = np.linspace(0, 1, N)
f = 10
A = 0.3

theta = A*np.sin(f * t)[None] + np.random.normal(size=(D, N))

B = 100
samples = np.random.normal(size=(B, D, N))
```

```python
plt.plot(theta[0])
```

```python
cv, ecp = tarp.get_drp_coverage(samples, theta)


plt.plot([0, 1], [0, 1], "k--", label="Calibrated", lw=3)
plt.plot(cv, ecp, "r-", label="Test", lw=3)
plt.legend()
plt.xlim(0, 1)
plt.ylim(0, 1)
```
